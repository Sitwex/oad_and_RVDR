# **задание 4 основные статистические показатели**

Среднее значение popularity и медиана popularity

```python
print("Среднее значение popularity:", df['popularity'].mean())
print("Медиана popularity:", df['popularity'].median())
```

Минимум, максимум и стандартное отклонение

```python
print("Минимум popularity:", df['popularity'].min())
print("Максимум popularity:", df['popularity'].max())
print("Стандартное отклонение popularity:", df['popularity'].std())
```

# **задание 7 гистограмма**

```python
# Создание фигуры нужного размера (чтобы график был крупным и читаемым)
plt.figure(figsize=(12, 6))

# Строю гистограмму для столбца 'popularity'
# bins=30 означает, что весь диапазон значений разобьется на 30 интервалов (столбиков)
plt.hist(df['popularity'], bins=30, color='skyblue', edgecolor='black')

# Добавление название графика
plt.title('Распределение значений popularity (популярности треков)', fontsize=14)

# Подпись ось X
plt.xlabel('Значение popularity', fontsize=12)

# Подпись ось Y
plt.ylabel('Количество треков (частота)', fontsize=12)

# Добавление сетки для удобства чтения
plt.grid(axis='y', alpha=0.75)

# Показ графика
plt.show()
```

ответы на вопросы
1. В каком диапазоне находится основная часть значений?
   Основная часть значений popularity сосредоточена в диапазоне от ~10 до 40. Именно здесь столбики гистограммы самые высокие, а пик приходится примерно на значение 20.
2. Есть ли значения, которые встречаются значительно реже остальных?
   Да, есть.
Слева: значения от 0 до ~5 встречаются редко.
Справа: значения выше 60 встречаются значительно реже. Особенно редкими являются треки с популярностью выше 80.
